```python
df["order_datetime"] = p.to_datetime(df["order_datetime"])
```

```python
df.dtypes
```

```python
df["Month"] = df["order_datetime"].dt.to_period("M").astype("string")
```

```python
df[["order_datetime", "Month"]]
```

```python
df["Day"] = df["order_datetime"].dt.day
df[["order_datetime", "Day"]]
```

```python
df["Weekday"] = df["order_datetime"].dt.day_name()
df[["order_datetime", "Weekday"]]
```

```python
category_summary = (
    df.groupby("category")
    .agg(
        TotalRevenue=("Revenue", "sum"),
        Orders=("order_id", "nunique"),
        AverageRevenuePerOrder=("Revenue", "mean")
    )
    .sort_values("TotalRevenue", ascending=False)
)

category_summary
```

```python
monthly_summary = (
    df.dropna(subset=["order_datetime"])
    .groupby("Month")
    .agg(
        TotalRevenue=("Revenue", "sum"),
        UnitsSold=("quantity", "sum"),
        Orders=("order_id", "nunique")
    )
    .sort_index()
)

monthly_summary
```

```python
m.figure(figsize=(8, 5))
category_summary["TotalRevenue"].plot(kind="bar", color="steelblue", title="Total Revenue by Category")
m.xlabel("category")
m.ylabel("Revenue")
m.xticks(rotation=45)
m.tight_layout()
m.show()
```

```python
m.figure(figsize=(8, 5))
monthly_summary["TotalRevenue"].plot(kind="line", marker="o", color="darkgreen", title="Monthly Revenue Trend")
m.xlabel("Month")
m.ylabel("Revenue")
m.xticks(rotation=45)
m.tight_layout()
m.show()
```

```python
m.figure(figsize=(8, 5))
category_summary["Orders"].plot(kind="bar", color="orange", title="Units Sold by Product Categpry")
m.xlabel("Product category")
m.ylabel("Units sold")
m.xticks(rotation=90)
m.tight_layout()
m.show()
```

```python
m.figure(figsize=(8, 5))
m.hist(df["Revenue"], bins=6, edgecolor="black")
m.title("Distribution of Revenue")
m.xlabel("Revenue")
m.ylabel("Frequency")
m.tight_layout()
m.show()
```

```python
m.figure(figsize=(8, 5))
m.scatter(df["quantity"], df["Revenue"], alpha=0.7)
m.title("Quantity vs Revenue")
m.xlabel("Quantity")
m.ylabel("Revenue")
m.tight_layout()
m.show()
```
