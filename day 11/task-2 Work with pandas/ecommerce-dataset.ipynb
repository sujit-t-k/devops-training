```python
df_no_dup["cost_per_unit"] = pd.to_numeric(df_no_dup["cost_per_unit"], errors = "coerce")
df_no_dup[["order_id", "order_datetime", "cost_per_unit"]]
```

```python
df_no_dup_no_cost_na = df_no_dup.dropna(subset = ["cost_per_unit", "order_datetime"])
df_no_dup_no_cost_na
```

```python
df_no_dup_no_cost_na_no_city_na = df_no_dup_no_cost_na.fillna("Los Vegas")
df_no_dup_no_cost_na_no_city_na
```
